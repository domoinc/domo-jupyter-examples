# Text Embeddings

`ai.embedding()` converts text into a numeric vector (embedding). Embeddings capture semantic meaning — texts with similar meanings have vectors that are close together. Use them for semantic search, similarity scoring, clustering, and recommendation.

## Embed a single string

Returns a raw dict with the embedding vector under `data[0]['embedding']`.

```python
import domojupyter.ai as ai

result = ai.embedding("Q3 revenue exceeded forecast by 12%")
vector = result["data"][0]["embedding"]
print(f"Embedding dimensions: {len(vector)}")
print(f"First 5 values: {vector[:5]}")
```

## Embed multiple strings at once

Pass a list to embed several texts in a single call.

```python
import domojupyter.ai as ai

texts = [
    "Revenue increased 15% year-over-year",
    "Customer churn rate dropped to 2.1%",
    "New product launch scheduled for Q4",
]

result = ai.embedding(texts)
print(f"{len(result['data'])} embeddings returned")
for i, item in enumerate(result["data"]):
    print(f"  Text {i}: {len(item['embedding'])} dimensions")
```

## Compute similarity between two texts

Cosine similarity ranges from -1 to 1. Values near 1 mean the texts are semantically similar.

```python
import domojupyter.ai as ai
import math

def cosine_similarity(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    norm_a = math.sqrt(sum(x ** 2 for x in a))
    norm_b = math.sqrt(sum(x ** 2 for x in b))
    return dot / (norm_a * norm_b)

result = ai.embedding([
    "How do I increase monthly recurring revenue?",
    "What are strategies to grow MRR?",
])

vec_a = result["data"][0]["embedding"]
vec_b = result["data"][1]["embedding"]
similarity = cosine_similarity(vec_a, vec_b)
print(f"Similarity: {similarity:.4f}")
```

## Semantic search

Embed a query and a list of candidates, then rank candidates by similarity to find the most relevant match.

```python
import domojupyter.ai as ai
import math

def cosine_similarity(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    norm_a = math.sqrt(sum(x ** 2 for x in a))
    norm_b = math.sqrt(sum(x ** 2 for x in b))
    return dot / (norm_a * norm_b)

query = "How to reduce employee turnover?"
candidates = [
    "Improving employee retention through better compensation",
    "Q3 product roadmap and feature releases",
    "Strategies for reducing staff attrition",
    "Sales pipeline management best practices",
    "Building a positive workplace culture to keep talent",
]

all_texts = [query] + candidates
result = ai.embedding(all_texts)

query_vec = result["data"][0]["embedding"]
scores = [
    (candidates[i], cosine_similarity(query_vec, result["data"][i + 1]["embedding"]))
    for i in range(len(candidates))
]
scores.sort(key=lambda x: x[1], reverse=True)

print(f"Query: {query}\n")
print("Ranked results:")
for text, score in scores:
    print(f"  [{score:.3f}] {text}")
```
