# Image Embeddings

`ai.image_embedding()` converts an image into a numeric vector. Use image embeddings for visual similarity search, image clustering, and cross-modal retrieval. Accepts local file paths or URLs.

## Embed a local image file

```python
import domojupyter.ai as ai

result = ai.image_embedding("product_photo.jpg")
vector = result["data"][0]["embedding"]
print(f"Embedding dimensions: {len(vector)}")
```

## Embed an image from a URL

```python
import domojupyter.ai as ai

result = ai.image_embedding("https://example.com/product_photo.jpg")
vector = result["data"][0]["embedding"]
print(f"Embedding dimensions: {len(vector)}")
```

## Compare two images

Cosine similarity near 1 means the images are visually similar.

```python
import domojupyter.ai as ai
import math

def cosine_similarity(a, b):
    dot = sum(x * y for x, y in zip(a, b))
    norm_a = math.sqrt(sum(x ** 2 for x in a))
    norm_b = math.sqrt(sum(x ** 2 for x in b))
    return dot / (norm_a * norm_b)

result_a = ai.image_embedding("image_a.jpg")
result_b = ai.image_embedding("image_b.jpg")

vec_a = result_a["data"][0]["embedding"]
vec_b = result_b["data"][0]["embedding"]

similarity = cosine_similarity(vec_a, vec_b)
print(f"Visual similarity: {similarity:.4f}")
```
