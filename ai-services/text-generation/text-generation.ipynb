# Text Generation

`ai.generate_text()` sends a prompt to a language model and returns generated text. Supports system messages, prompt templates, structured output (JSON), and reasoning.

## Basic prompt

```python
import domojupyter.ai as ai

response = ai.generate_text("Summarize the key benefits of cloud computing in three bullet points.")
print(response.choices[0])
```

## With a system message

Use `system` to set a persona or constrain how the model responds.

```python
import domojupyter.ai as ai

response = ai.generate_text(
    "What should I focus on in my Q3 business review?",
    system="You are a concise business analyst. Respond in plain English with no jargon. Limit your answer to 5 sentences."
)
print(response.choices[0])
```

## With a prompt template and parameters

Templates let you reuse a prompt structure with different variable values.

```python
import domojupyter.ai as ai

template = ai.PromptTemplate("Explain ${topic} to a ${audience} in ${num_sentences} sentences or fewer.")
params = {
    "topic": "machine learning",
    "audience": "non-technical executive",
    "num_sentences": "3"
}

response = ai.generate_text("", prompt_template=template, parameters=params)
print(response.choices[0])
```

## With a specific model

Override the default model by passing its ID.

```python
import domojupyter.ai as ai

response = ai.generate_text(
    "Draft a one-paragraph executive summary for a Q3 revenue report showing 12% growth.",
    model="your-model-id"
)
print(response.choices[0])
```

## With temperature and max tokens

`temperature` (0.0–1.0) controls randomness. Lower values produce more deterministic output. `maxTokens` caps the response length.

```python
import domojupyter.ai as ai

response = ai.generate_text(
    "Write a creative product tagline for a data analytics platform.",
    temperature=0.9,
    maxTokens=50
)
print(response.choices[0])
```

## With structured output (response_format)

Request a JSON response conforming to a schema. Useful when you need to parse the output programmatically.

```python
import domojupyter.ai as ai
import json

schema = {
    "properties": {
        "summary": {"type": "string"},
        "key_risks": {"type": "array", "items": {"type": "string"}},
        "recommendation": {"type": "string"}
    },
    "required": ["summary", "key_risks", "recommendation"]
}

response = ai.generate_text(
    "Analyze the following scenario: Sales are up 15% but customer churn increased by 8% this quarter.",
    response_format={"type": "JSON", "schema": schema}
)

result = json.loads(response.choices[0])
print("Summary:", result["summary"])
print("Risks:", result["key_risks"])
print("Recommendation:", result["recommendation"])
```

## With extended reasoning (reasoning_config)

Enable the model to think step-by-step before answering. Best for complex analytical or multi-step problems. `budgetTokens` controls how much thinking is allowed.

```python
import domojupyter.ai as ai

response = ai.generate_text(
    "A company has $2M to invest. Option A yields 8% annually with low risk. Option B yields 15% with high risk. "
    "Option C is a 50/50 split. Assuming a 5-year horizon and 30% tax rate, which option maximizes after-tax returns?",
    reasoning_config={"enabled": True, "budgetTokens": 2000}
)
print(response.choices[0])
```
