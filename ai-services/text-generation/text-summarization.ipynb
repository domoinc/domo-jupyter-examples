# Text Summarization

`ai.summarize()` condenses long text into a shorter form. Control the output style, word count, and how the input is chunked for large documents.

## Basic summarization

```python
import domojupyter.ai as ai

article = """
Global supply chains faced significant disruption in 2023 due to geopolitical tensions, rising energy costs, 
and ongoing labor shortages in key manufacturing regions. Companies responded by diversifying suppliers, 
increasing safety stock, and accelerating investment in automation. While short-term costs rose, 
many organizations reported improved resilience and reduced single-source dependency by year-end.
"""

response = ai.summarize(article)
print(response.choices[0])
```

## Control output format

Use `output_style` to get a bulleted list or paragraph summary.

```python
import domojupyter.ai as ai

text = "Your long document text goes here..."

# BULLETED produces a bullet-point list; PARAGRAPH produces flowing prose
response = ai.summarize(text, output_style=ai.SummarizationOutputStyle.BULLETED)
print(response.choices[0])
```

## Control output length

Use `output_word_length` with a `SizeBoundary` to set min and max word counts.

```python
import domojupyter.ai as ai

text = "Your long document text goes here..."

response = ai.summarize(
    text,
    output_style=ai.SummarizationOutputStyle.PARAGRAPH,
    output_word_length=ai.SizeBoundary(100, 150)
)
print(response.choices[0])
```

## Handle large documents with chunking

For documents that exceed the model's context window, use `ChunkingConfiguration` to split the text into overlapping chunks that are summarized individually and then combined.

```python
import domojupyter.ai as ai

large_document = "Your large document text goes here..."

# ChunkingConfiguration(max_chunk_size, chunk_overlap, separators, separator_type, disallow_intermediate_chunks)
chunking = ai.ChunkingConfiguration(
    max_chunk_size=500,
    chunk_overlap=50,
    list_of_separators=["\n\n", "\n", ". "],
    separator_type="TEXT",
    disallow_intermediate_chunks=False
)

response = ai.summarize(
    large_document,
    chunking_configuration=chunking,
    output_style=ai.SummarizationOutputStyle.BULLETED,
    output_word_length=ai.SizeBoundary(150, 300)
)
print(response.choices[0])
```

## With a custom prompt template

Override the default summarization instructions with your own template.

```python
import domojupyter.ai as ai

text = "Your document text goes here..."

template = ai.PromptTemplate(
    "You are a financial analyst. Summarize the following text focusing on monetary figures, "
    "risks, and opportunities. Limit the summary to ${max_words} words.\n\nText: ${input}"
)

response = ai.summarize(
    text,
    prompt_template=template,
    parameters={"max_words": "200"}
)
print(response.choices[0])
```
