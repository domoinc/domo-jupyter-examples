# Text to Beast Mode

`ai.text_to_beast_mode()` generates a Domo Beast Mode (calculated field) formula from a plain English description. Optionally provide a data source schema so the model knows your column names.

## Basic formula from a description

No schema needed for simple calculations that don't reference specific column names.

```python
import domojupyter.ai as ai

response = ai.text_to_beast_mode("Calculate profit margin as a percentage")
print(response.choices[0])
```

## With a data source schema

Provide column names and types so the formula references your actual fields.

```python
import domojupyter.ai as ai

schema = ai.DataSourceSchema(
    "Sales",
    "Monthly sales transactions",
    [
        ai.Column("Revenue",    "DOUBLE"),
        ai.Column("COGS",       "DOUBLE"),
        ai.Column("Region",     "STRING"),
        ai.Column("Date",       "DATE"),
        ai.Column("Sales_Rep",  "STRING"),
    ]
)

response = ai.text_to_beast_mode(
    "Show gross profit margin as a percentage, using Revenue and COGS columns",
    data_source_schema=schema
)
print(response.choices[0])
```

## Conditional / CASE formula

Describe a conditional rule in plain English to generate an IF/CASE expression.

```python
import domojupyter.ai as ai

schema = ai.DataSourceSchema(
    "Deals",
    "Sales pipeline deals",
    [
        ai.Column("Deal_Size",   "DOUBLE"),
        ai.Column("Close_Date",  "DATE"),
        ai.Column("Stage",       "STRING"),
        ai.Column("Sales_Rep",   "STRING"),
    ]
)

response = ai.text_to_beast_mode(
    "Categorize deals as 'Small' if under $10k, 'Medium' if $10k-$100k, and 'Large' if over $100k",
    data_source_schema=schema
)
print(response.choices[0])
```

## With a specific model

```python
import domojupyter.ai as ai

schema = ai.DataSourceSchema(
    "Finance",
    "Financial metrics",
    [
        ai.Column("Actual",   "DOUBLE"),
        ai.Column("Budget",   "DOUBLE"),
        ai.Column("Forecast", "DOUBLE"),
    ]
)

response = ai.text_to_beast_mode(
    "Calculate variance between Actual and Budget as both an absolute value and a percentage",
    data_source_schema=schema,
    model="your-model-id"
)
print(response.choices[0])
```
