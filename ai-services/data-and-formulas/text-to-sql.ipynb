# Text to SQL

`ai.text_to_sql()` converts a natural language question into a SQL query. Provide a schema so the model knows your column names and types. You can pass a schema definition directly, a connected Domo dataset alias, or a pandas DataFrame.

## Define a schema inline

Use `DataSourceSchema` and `Column` when you want full control over what the model sees. Supported column types: `STRING`, `DOUBLE`, `LONG`, `DECIMAL`, `DATE`, `DATETIME`.

```python
import domojupyter.ai as ai

schemas = [
    ai.DataSourceSchema(
        "Sales",
        "Monthly sales transactions",
        [
            ai.Column("Date",       "DATE"),
            ai.Column("Region",     "STRING"),
            ai.Column("Product",    "STRING"),
            ai.Column("Revenue",    "DOUBLE"),
            ai.Column("Units_Sold", "LONG"),
        ]
    )
]

response = ai.text_to_sql(
    "What is total revenue by region for the last 6 months?",
    data_source_schemas=schemas
)
print(response.choices[0])
```

## Use a connected Domo dataset

If the dataset is already connected to the workspace, pass its alias directly. The service will infer the schema automatically.

```python
import domojupyter.ai as ai

response = ai.text_to_sql(
    "Show me the top 10 customers by revenue year-to-date",
    workspace_data_source_alias="your_dataset_name"
)
print(response.choices[0])
```

## Use a pandas DataFrame

Read data into a DataFrame first (useful when you've transformed the data or it's not a direct dataset alias).

```python
import domojupyter.ai as ai
import domojupyter as domo

df = domo.read_dataframe("your_dataset_name", query="SELECT * FROM table")

response = ai.text_to_sql(
    "Which products had declining sales month-over-month?",
    dataframe=df
)
print(response.choices[0])
```

## Multi-table query

Pass multiple `DataSourceSchema` objects when your question spans more than one table.

```python
import domojupyter.ai as ai

schemas = [
    ai.DataSourceSchema(
        "Orders",
        "Customer orders",
        [
            ai.Column("Order_ID",    "STRING"),
            ai.Column("Customer_ID", "STRING"),
            ai.Column("Order_Date",  "DATE"),
            ai.Column("Revenue",     "DOUBLE"),
        ]
    ),
    ai.DataSourceSchema(
        "Customers",
        "Customer master data",
        [
            ai.Column("Customer_ID",  "STRING"),
            ai.Column("Name",          "STRING"),
            ai.Column("Segment",       "STRING"),
            ai.Column("Country",       "STRING"),
        ]
    ),
]

response = ai.text_to_sql(
    "What is total revenue by customer segment and country?",
    data_source_schemas=schemas
)
print(response.choices[0])
```

## With a specific model

Override the default model by specifying a model ID.

```python
import domojupyter.ai as ai

response = ai.text_to_sql(
    "Calculate the 30-day rolling average revenue per user",
    workspace_data_source_alias="your_dataset_name",
    model="your-model-id"
)
print(response.choices[0])
```
