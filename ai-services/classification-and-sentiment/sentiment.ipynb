# Sentiment Analysis

`ai.sentiment()` analyzes the emotional tone of text.

- **Without topics**: returns overall sentiment for the full text
- **With topics** (`SentimentTopic` list): returns targeted sentiment per topic — the model identifies how the text feels about each specific aspect

## General sentiment analysis

Analyze the overall tone of a piece of text.

```python
import domojupyter.ai as ai

review = """
The product works well overall, but the onboarding took way longer than expected 
and the pricing feels steep for what you get. Customer support was responsive though.
"""

result = ai.sentiment(review)
print(result)
```

## Targeted sentiment by topic

Pass `SentimentTopic` objects to get per-topic sentiment scores. The model analyzes how the text feels about each specific aspect. Topic names must start with a letter and contain only letters, numbers, and underscores.

```python
import domojupyter.ai as ai

topics = [
    ai.SentimentTopic("price"),
    ai.SentimentTopic("quality"),
    ai.SentimentTopic("support"),
    ai.SentimentTopic("onboarding"),
]

review = """
The product works well overall, but the onboarding took way longer than expected 
and the pricing feels steep for what you get. Customer support was responsive though.
"""

result = ai.sentiment(review, topics=topics)
print(result)
```

## With topic descriptions

Add `description` to clarify what each topic represents, especially when the name is ambiguous or covers a broad concept.

```python
import domojupyter.ai as ai

topics = [
    ai.SentimentTopic("price",     "Cost, pricing plans, value for money, or affordability"),
    ai.SentimentTopic("quality",   "Product functionality, reliability, and performance"),
    ai.SentimentTopic("support",   "Customer service, responsiveness, and helpfulness of support staff"),
    ai.SentimentTopic("onboarding", "Setup process, documentation, and time to first value"),
]

review = """Getting started was a nightmare — the docs are outdated and nobody replied for 3 days.
          Once I got going, the product itself is solid and worth every penny."""

result = ai.sentiment(review, topics=topics)
print(result)
```

## With instructions

Use `instructions` to customize how the model performs the analysis.

```python
import domojupyter.ai as ai

topics = [
    ai.SentimentTopic("product"),
    ai.SentimentTopic("company"),
]

text = "The app itself is great but I don't trust this company after the data breach."

result = ai.sentiment(
    text,
    topics=topics,
    instructions="Distinguish clearly between sentiment about the product and sentiment about the company as an organization."
)
print(result)
```

## Analyze multiple reviews in a loop

Practical pattern for processing a batch of reviews.

```python
import domojupyter.ai as ai

reviews = [
    "Fantastic product, highly recommend to everyone!",
    "It works okay but has too many bugs for the price.",
    "Terrible support. Waited a week and never got a resolution.",
    "Exactly what we needed. Setup was easy and the team loved it.",
]

for review in reviews:
    result = ai.sentiment(review)
    print(f"{review[:60]}...")
    print(f"  => {result}\n")
```
