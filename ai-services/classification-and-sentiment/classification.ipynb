# Classification

`ai.classification()` assigns one or more labels to a piece of text. Define the labels you care about and the model scores the input against them.

- **SINGLE_LABEL**: the model picks the best-fitting label
- **MULTI_LABEL**: the model can assign multiple labels

## Single-label classification

Classify a customer message into exactly one support category.

```python
import domojupyter.ai as ai

labels = [
    ai.ClassificationLabel("billing"),
    ai.ClassificationLabel("technical"),
    ai.ClassificationLabel("account_access"),
    ai.ClassificationLabel("general_inquiry"),
]

message = "I was charged twice for my subscription this month and need a refund."

result = ai.classification(
    message,
    labels=labels,
    classification_type="SINGLE_LABEL"
)
print(result)
```

## Multi-label classification

A single text can belong to multiple categories. `MULTI_LABEL` allows the model to assign more than one.

```python
import domojupyter.ai as ai

labels = [
    ai.ClassificationLabel("finance"),
    ai.ClassificationLabel("technology"),
    ai.ClassificationLabel("healthcare"),
    ai.ClassificationLabel("sustainability"),
    ai.ClassificationLabel("mergers_and_acquisitions"),
]

headline = "Tech giant acquires health data startup to expand AI-driven diagnostics platform."

result = ai.classification(
    headline,
    labels=labels,
    classification_type="MULTI_LABEL"
)
print(result)
```

## With label descriptions

Add `description` to each label when the label names alone might be ambiguous. This helps the model distinguish between similar categories.

```python
import domojupyter.ai as ai

labels = [
    ai.ClassificationLabel("positive",  "Customer is satisfied, expressing appreciation or approval"),
    ai.ClassificationLabel("neutral",   "Customer is factual, asking a question, or making a request without emotional tone"),
    ai.ClassificationLabel("negative",  "Customer is frustrated, dissatisfied, or complaining"),
    ai.ClassificationLabel("escalation", "Customer is threatening to cancel, leave a review, or contact management"),
]

feedback = "This is the third time I've had to contact support for the same issue. I'm losing patience."

result = ai.classification(
    feedback,
    labels=labels,
    classification_type="SINGLE_LABEL"
)
print(result)
```

## With instructions

Use `instructions` to give the model additional guidance about edge cases or classification logic.

```python
import domojupyter.ai as ai

labels = [
    ai.ClassificationLabel("urgent"),
    ai.ClassificationLabel("normal"),
    ai.ClassificationLabel("low_priority"),
]

ticket = "Hey, when you get a chance, could you update my email address on file?"

result = ai.classification(
    ticket,
    labels=labels,
    classification_type="SINGLE_LABEL",
    instructions="Classify based on business impact and time sensitivity. "
               "Account access and security issues are always urgent. "
               "Profile updates with no urgency language are low_priority."
)
print(result)
```

## Classify multiple texts in a loop

Practical pattern for processing a batch of records.

```python
import domojupyter.ai as ai

labels = [
    ai.ClassificationLabel("billing"),
    ai.ClassificationLabel("technical"),
    ai.ClassificationLabel("account_access"),
    ai.ClassificationLabel("general_inquiry"),
]

tickets = [
    "I can't log into my account after resetting my password.",
    "What are your business hours?",
    "The dashboard keeps showing a 500 error when I click on Reports.",
    "I need a copy of my invoice from last month.",
]

for ticket in tickets:
    result = ai.classification(ticket, labels=labels, classification_type="SINGLE_LABEL")
    print(f"[{ticket[:60]}...] => {result}")
```
