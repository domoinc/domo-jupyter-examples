# Image to Text

`ai.image_to_text()` analyzes an image and returns a text description or answer. Use it to describe images, extract information, or pull structured data from charts, documents, and forms. Accepts local file paths or URLs.

## Describe an image (local file)

No prompt needed — the model generates a description by default.

```python
import domojupyter.ai as ai

response = ai.image_to_text("chart.png")
print(response.choices[0])
```

## Ask a specific question about an image

Use `input_str` to ask the model something specific about the image content.

```python
import domojupyter.ai as ai

response = ai.image_to_text(
    "sales_chart.png",
    input_str="What is the trend in the data? Which month had the highest value?"
)
print(response.choices[0])
```

## From a URL

```python
import domojupyter.ai as ai

response = ai.image_to_text(
    "https://example.com/dashboard_screenshot.png",
    input_str="List the KPIs shown in this dashboard and their current values."
)
print(response.choices[0])
```

## Extract structured data from an image

Use `response_format` to pull specific fields from a document, form, or chart image into a structured JSON object.

```python
import domojupyter.ai as ai
import json

schema = {
    "properties": {
        "invoice_number": {"type": "string"},
        "vendor_name": {"type": "string"},
        "total_amount": {"type": "number"},
        "due_date": {"type": "string"}
    },
    "required": ["invoice_number", "vendor_name", "total_amount", "due_date"]
}

response = ai.image_to_text(
    "invoice.jpg",
    input_str="Extract the invoice details from this document.",
    response_format={"type": "JSON", "schema": schema}
)

data = json.loads(response.choices[0])
print(f"Invoice #: {data['invoice_number']}")
print(f"Vendor:    {data['vendor_name']}")
print(f"Amount:    ${data['total_amount']}")
print(f"Due:       {data['due_date']}")
```

## Analyze a PDF document

Pass a `.pdf` file path to extract text, answer questions, or pull structured data from a document.

```python
import domojupyter.ai as ai

response = ai.image_to_text(
    "report.pdf",
    input_str="Summarize the key findings from this document."
)
print(response.choices[0])
```
